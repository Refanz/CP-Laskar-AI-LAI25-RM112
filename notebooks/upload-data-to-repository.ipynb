# **Upload Dataset to Dagshub**

## **Import Libraries**

```python
import dagshub
from dagshub.data_engine import datasources
from dagshub import upload_files
```

## **Init Dagshub**

```python
dagshub.init(repo_owner='refandasuryasaputra', repo_name='CP-Laskar-AI-LAI25-RM112', mlflow=True)
```

## **Upload Dataset to Dagshub**

```python
data_path = "../palm-disease-dataset/"

upload_files('refandasuryasaputra/CP-Laskar-AI-LAI25-RM112', data_path)
```

## **Create a DataSource from Dagshub**

```python
data_path = "../palm-disease-dataset/"

ds = datasources.create_datasource('refandasuryasaputra/CP-Laskar-AI-LAI25-RM112', 'palm-datasets', data_path)
ds.head().dataframe
```
